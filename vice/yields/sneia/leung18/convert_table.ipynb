```python
import numpy as np
import pandas as pd
import vice
```

```python
table = pd.read_csv("leung18_table4.txt", sep="\t", comment="#")
table
```

```python
["zn" + num for num in table[table["Isotopes"].str.contains("Zn")]["Isotopes"].str[1:3]]
```

```python
table[table["Isotopes"].str.contains("Zn")]["Z = 0.1 Z_sun"]
```

```python
for Z in [0, 0.1, 0.5, 1, 2, 3, 5]:
    col = "Z = %i Z_sun" % Z
    model = "300-%s-C3-1" % str(Z).replace('.', 'P')
    for el in vice.elements.recognized:
        filename = '%s/%s.dat' % (model, el)
        print(filename)
        element_mask = table["Isotopes"].str.contains(el.capitalize())
        isotopes = [el + num for num in table[element_mask]["Isotopes"].str[1:3]]
        yields = table[element_mask][col].astype("float64")
        with open(filename, 'w') as f:
            f.write("# isotope	Mass yield (Msun)\n")
            f.writelines([f"{i}\t{y:.02e}\n" for i, y in zip(isotopes, yields)])
            f.write("\n")
```
